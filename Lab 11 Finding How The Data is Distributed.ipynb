### Step 3: Handle Missing Data


- Identify missing values in the dataset.

- Impute or remove missing values as necessary to ensure data completeness.



```python
## Write your code here
missing_values = df.isnull().sum().sort_values(ascending=False)
missing_values
```

```python
# drop a subset of columns
remove = ['AINextMuch less integrated', 'AINextLess integrated','AINextNo change','AINextMuch more integrated','EmbeddedAdmired',
         'JobSatPoints_11','JobSatPoints_4','JobSatPoints_5','JobSatPoints_6','JobSatPoints_7','JobSatPoints_8','JobSatPoints_9','JobSatPoints_10']

df = df.drop(columns=remove)
df.head()
```

```python
# STandardize the values for Educational level
replace_Values = {
    "Bachelor’s degree (B.A., B.S., B.Eng., etc.)": "Bachelor's Degree", 
    "Master’s degree (M.A., M.S., M.Eng., MBA, etc.)": "Master’s Degree",
    "Some college/university study without earning a degree":"Undergraduate Studies (Incomplete)",
    "Professional degree (JD, MD, Ph.D, Ed.D, etc.)":"Professional Degree",
    "Associate degree (A.A., A.S., etc.)":"Associate Degree",
    "Secondary school (e.g. American high school, German Realschule or Gymnasium, etc.)":"Secondary School",
    "Something else":"Other"
}
df['EdLevel'] = df['EdLevel'].replace(replace_Values) 
#pd.set_option('display.max_rows', None) # show all rows in output
df['EdLevel'].value_counts()
#impute missing values by mean and most frequent value
frequent = df['RemoteWork'].mode()[0]
frequentJobSat = df['JobSat'].mode()[0]
mean_value = df['ConvertedCompYearly'].mean()
df['RemoteWork'].fillna(frequent, inplace=True)
df['JobSat'].fillna(frequentJobSat, inplace=True)
df['ConvertedCompYearly'].fillna(mean_value, inplace=True)
```

```python
missing_values = df.isnull().sum().sort_values(ascending=False)
missing_values
```

```python
# function to impute the remaining missing values with mode

def fill_missing_with_mode(df):
    for column in df.columns:
        mode_series = df[column].mode()
        if not mode_series.empty:
            mode_value = mode_series[0]
            df[column] = df[column].fillna(mode_value)
            print(f"filled missing values in '{column}'with mode: {mode_value}")
        else:
            print(f"'{column}' has no mode (possibly all values are NaN)")
    return df


df = fill_missing_with_mode(df)
```

```python
## Write your code here

def calculate_value_counts(df):
    columns_of_interest = ['Employment', 'JobSat', 'YearsCodePro']
    value_counts = {}
    
    for column in columns_of_interest:
        if column in df.columns:
            value_counts[column] = df[column].value_counts(dropna=False)
        else:
            print(f"Column '{column}' not found in DataFrame.")
    
    return value_counts



counts = calculate_value_counts(df)
for col, count in counts.items():
    print(f"\nValue counts for {col}:\n{count}")
```

### Step 7: Analyze Remote Work Trends


- Visualize the distribution of RemoteWork by region using a grouped bar chart or heatmap.


```python
## Write your code here
!pip install pycountry-convert
import pycountry_convert as pc

# Function to map country name to continent
def get_continent(country_name):
    try:
        # Convert country name to ISO Alpha-2 code
        country_code = pc.country_name_to_country_alpha2(country_name)
        # Convert Alpha-2 code to continent code
        continent_code = pc.country_alpha2_to_continent_code(country_code)
        # Map continent code to continent name
        continent_name = {
            'AF': 'Africa',
            'AS': 'Asia',
            'EU': 'Europe',
            'NA': 'North America',
            'SA': 'South America',
            'OC': 'Oceania'
        }
        return continent_name.get(continent_code, 'Unknown')
    except:
        return 'Unknown'

# Apply the function to create a Region column
df['Region'] = df['Country'].apply(get_continent)

```

```python

# Count occurrences of RemoteWork by Region
grouped = df.groupby(['Region', 'RemoteWork']).size().unstack(fill_value=0)

# Plot grouped bar chart
grouped.plot(kind='bar', figsize=(8, 6))
plt.title('Distribution of Remote Work by Region')
plt.xlabel('Region')
plt.ylabel('Number of Respondents')
plt.xticks(rotation=45)
plt.legend(title='Remote Work Type')
plt.tight_layout()
plt.show()
```
