The functions below will download the dataset into your browser:



```python
# Define the URL of the dataset
file_path = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/n01PQ9pSmiRX6520flujwQ/survey-data.csv"

# Load the dataset into a DataFrame
df = pd.read_csv(file_path)

# Display the first few rows to ensure it loaded correctly
print(df.head())
```

### 2. Explore the Dataset
##### Task 1: Display basic information and summary statistics of the dataset.


```python
## Write your code here
df.describe()
```

### 3. Finding Missing Values
##### Task 2: Identify missing values for all columns.


```python
## Write your code here
missing_values = df.isnull().sum()
missing_values
```

##### Task 3: Visualize missing values using a heatmap (Using seaborn library).



```python
## Write your code here
plt.figure(figsize=(14, 7))
sns.heatmap(df.isnull(), cbar=False, cmap='viridis', yticklabels=False)
plt.title('Heatmap of Missing Values')
plt.xlabel('Columns')
```

##### Task 4: Count the number of missing rows for a specific column (e.g., `Employment`).


```python
## Write your code here
num_missing_CCY = df['ConvertedCompYearly'].isnull().sum()
print(f"Number of missing rows in 'ConvertedCompYearly' column: {num_missing_CCY}")
```

### 4. Imputing Missing Values
##### Task 5: Identify the most frequent (majority) value in a specific column (e.g., `Employment`).


```python
## Write your code here
most_frequent_values=df['AIThreat'].mode()[0]
print(f"The most freqquent value in 'AIThreat' column is: {most_frequent_values}")
```

##### Task 6: Impute missing values in the `Employment` column with the most frequent value.



```python
## Write your code here
most_frequent_value_employ=df['Employment'].mode()[0]
df['Employment'].fillna(most_frequent_value_employ, inplace=True)
print(f"The missing values in 'Employment' column has been replaced with: {most_frequent_value_employ}")
```

### 5. Visualizing Imputed Data
##### Task 7: Visualize the distribution of a column after imputation (e.g., `Employment`).


```python
## Write your code here
employment_counts = df['Employment'].value_counts()

plt.figure(figsize=(100, 100)) 
employment_counts.plot(kind= 'pie', autopct=lambda pct: f'{pct:.10f}%', startangle=140, colors=plt.cm.Paired.colors, textprops={'fontsize': 20})
plt.title('Distribution of Employment Status')
plt.ylabel('')
plt.tight_layout()
plt.show
```
