### 1. Load the Dataset


<h5>1.1 Import necessary libraries and load the dataset.</h5>


Ensure the dataset is loaded correctly by displaying the first few rows.


```python
# Import necessary libraries
import pandas as pd

# Load the Stack Overflow survey data
dataset_url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/n01PQ9pSmiRX6520flujwQ/survey-data.csv"
df = pd.read_csv(dataset_url)

# Display the first few rows
print(df.head())
```

#### 2. Explore the Dataset


<h5>2.1 Summarize the dataset by displaying the column data types, counts, and missing values.</h5>


```python
# Write your code here
data_types = df.dtypes
num_col = df.shape[1]
missing_values = df.isnull().sum()
print(f" the types of column data types are: {data_types}")
print("\n")
print(f"the number of columns are: {num_col}")
print("\n")
print(f"the missing values in the dataset are: {missing_values}")
```

<h5>2.2 Generate basic statistics for numerical columns.</h5>


```python
# Write your code here
df.describe()
```

### 3. Identifying and Removing Inconsistencies


<h5>3.1 Identify inconsistent or irrelevant entries in specific columns (e.g., Country).</h5>


```python
pd.set_option('display.max_rows', None) # show all rows in output
df['Country'].value_counts()
```

```python
# Write your code here
df['Country'] = df['Country'].replace('Viet Nam', 'Vietnam')
pd.set_option('display.max_rows', None) # show all rows in output
df['Country'].value_counts()
```

<h5>3.2 Standardize entries in columns like Country or EdLevel by mapping inconsistent values to a consistent format.</h5>


```python
pd.set_option('display.max_rows', None) # show all rows in output
df['EdLevel'].value_counts()
```

```python
## Write your code here
replace_Values = {
    "Bachelor’s degree (B.A., B.S., B.Eng., etc.)": "Bachelor's Degree", 
    "Master’s degree (M.A., M.S., M.Eng., MBA, etc.)": "Master’s Degree",
    "Some college/university study without earning a degree":"Undergraduate Studies (Incomplete)",
    "Professional degree (JD, MD, Ph.D, Ed.D, etc.)":"Professional Degree",
    "Associate degree (A.A., A.S., etc.)":"Associate Degree",
    "Secondary school (e.g. American high school, German Realschule or Gymnasium, etc.)":"Secondary School",
    "Something else":"Other"
}

df['EdLevel'] = df['EdLevel'].replace(replace_Values) 
pd.set_option('display.max_rows', None) # show all rows in output
df['EdLevel'].value_counts()
```

### 4. Encoding Categorical Variables


<h5>4.1 Encode the Employment column using one-hot encoding.</h5>


```python
df['Employment'].value_counts()
```

```python
## Write your code here
df_encoded = pd.get_dummies(df, columns=['Employment'])
print(df_encoded.head())
```

### 5. Handling Missing Values


<h5>5.1 Identify columns with the highest number of missing values.</h5>


```python
## Write your code here
missing_values = df.isnull().sum().sort_values(ascending=False)
missing_values
```

<h5>5.2 Impute missing values in numerical columns (e.g., `ConvertedCompYearly`) with the mean or median.</h5>


```python
## Write your code here
mean_comp = df['ConvertedCompYearly'].mean()
df['ConvertedCompYearly'].fillna(mean_comp, inplace=True)
```

<h5>5.3 Impute missing values in categorical columns (e.g., `RemoteWork`) with the most frequent value.</h5>


```python
## Write your code here
most_frequent = df['RemoteWork'].mode()[0]
df['RemoteWork'].fillna(most_frequent)
print(f"the missing value in the RemoteWork column has been replaced with: {most_frequent}")
```

### 6. Feature Scaling and Transformation


<h5>6.1 Apply Min-Max Scaling to normalize the `ConvertedCompYearly` column.</h5>


```python
## Write your code here
min_val = df['ConvertedCompYearly'].min()

# Find the maximum value in the column
max_val = df['ConvertedCompYearly'].max()

# Apply Min-Max Scaling formula to normalize the values
# Formula: (value - min) / (max - min)
df['ConvertedCompYearly_MinMax'] = (df['ConvertedCompYearly'] - min_val) / (max_val - min_val)

# Display the DataFrame with the new normalized column
print(df.head(5))
```

<h5>6.2 Log-transform the ConvertedCompYearly column to reduce skewness.</h5>


```python
## Write your code here
import numpy as np
# Apply log transformation to reduce skewness
df['ConvertedCompYearly_Log'] = np.log1p(df['ConvertedCompYearly'])

# Display the DataFrame with the new log-transformed column
print(df.head(5))
```

### 7. Feature Engineering


<h5>7.1 Create a new column `ExperienceLevel` based on the `YearsCodePro` column:</h5>


```python
## Write your code here

# Convert 'YearsCodePro' to numeric values
def convert_years(value):
    if value == 'Less than 1 year':
        return 0.5
    else:
        try:
            return float(value)
        except ValueError:
            return None

df['YearsCodePro_Num'] = df['YearsCodePro'].apply(convert_years)

# Categorize experience levels
def categorize_experience(years):
    if years is None:
        return 'Unknown'
    elif years < 2:
        return 'Beginner'
    elif years < 5:
        return 'Intermediate'
    elif years < 10:
        return 'Advanced'
    else:
        return 'Expert'

df['ExperienceLevel'] = df['YearsCodePro_Num'].apply(categorize_experience)

# Display the result
print(df.head())
```
